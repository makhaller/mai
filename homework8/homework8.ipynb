```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

print(median_lat, median_lon)
```

```python
df["lat_diff"] = abs(df["latitude"] - median_lat)
df["lon_diff"] = abs(df["longitude"] - median_lon)

df
```

```python
#closest to median latitude 
closest_lat = df.loc[df["lat_diff"].idxmin()]

#farthest from median latitude
farthest_lat = df.loc[df["lat_diff"].idxmax()]

#closest to median longitude 
closest_lon = df.loc[df["lon_diff"].idxmin()]

#farthest to median longitude 
farthest_lon = df.loc[df["lon_diff"].idxmax()]

print(f"Closest to median latitude: {closest_lat['state']} ({closest_lat['lat_diff']:.2f})")
print(f"Farthest from median latitude: {farthest_lat['state']} ({farthest_lat['lat_diff']:.2f})")

print(f"Closest to median longitude: {closest_lon['state']} ({closest_lon['lon_diff']:.2f})")
print(f"Farthest from median longitude: {farthest_lon['state']} ({farthest_lon['lon_diff']:.2f})")
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df2 = sns.load_dataset("planets")
df2
```

```python
df2 = df2.dropna()
df2
```

```python
print("Shape:", df2.shape)
print("Rows:", df2.shape[0])
print("Columns:", df2.shape[1])
```

```python
df2.head()
```

```python
df2.tail()
```

```python
print(df2.columns)

#method = discovery method used to find the planet 
#number = number of planets in the planetary system
#orbit period = the time the planet takes to orbit its star in days 
#mass = planets mass, measured in Jupiter masses
#distance = distance from Earth in light years
#year = year the planet was discovered 
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Top-left scatterplot
sns.scatterplot(
    data=df2,
    x="orbital_period",
    y="mass",
    hue="method",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Orbital Period vs Mass")
axes[0, 0].set_xlabel("Orbital Period (days)")
axes[0, 0].set_ylabel("Mass (Jupiter masses)")

# Top-right scatterplot
sns.scatterplot(
    data=df2,
    x="distance",
    y="mass",
    hue="method",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Distance vs Mass")
axes[0, 1].set_xlabel("Distance (light years)")
axes[0, 1].set_ylabel("Mass (Jupiter masses)")

# Bottom-left histogram
sns.histplot(
    data=df2,
    x="year",
    ax=axes[1, 0],
    color="orange"
)
axes[1, 0].set_title("Distribution of Discovery Year")
axes[1, 0].set_xlabel("Year")
axes[1, 0].set_ylabel("Count")

# Bottom-right boxplot
sns.boxplot(
    data=df2,
    x="method",
    y="mass",
    ax=axes[1, 1],
    color="lightgreen"
)
axes[1, 1].set_title("Mass by Discovery Method")
axes[1, 1].set_xlabel("Discovery Method")
axes[1, 1].set_ylabel("Mass (Jupiter masses)")

plt.show()
```

```python
#What the data represents and why I decided on these columns 
#Top-left = this scatterplot shows the relationship between the orbital period of a planet and its mass. I chose these columns to see whether planets with longer orbitals are usually more or less massive
#Top-right = this scatterplot shows the relationship between the distance from the Earth and the planet's mass. I chose these columns to compare how far away planets are and whether the planet's mass varies with distance 
#Bottom-left = this histogram shows the distribution of discovery years of the planets. I chose this column to see when the most planets in the dataset were discovered
#Bottom-right = this boxplot compares planet mass across discovery methods. I chose this ot see whether some methods tend to detect larger or smaller planets

#What I learned 
#Top-left = the orbital periods vary a lot, and there does not appear to be a single simple relationship between orbital period and mass (i.e., orbital period and mass seem to not be correlated)
#Top-right = distances also vary widely, and planets of many masses are found at different distances. One distance/area does not seem to "make" more massive planets 
#Bottom-left = more planets were discovered in later years, showing a growth in exoplanet detection over time 
#Bottom-right = different discovery methods appear to detect planets with different mass ranges, it seems that the method affects what kinds of planets are easiest to find 
```
