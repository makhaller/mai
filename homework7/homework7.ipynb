```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w): 
    return A * np.sin(w * x)

x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axs = plt.subplots(1, 3, figsize=(25, 5))

for A in amplitude:
    y = make_sine_wave(x, A, 1) 
    axs[0].plot(x, y, label=f"A={A}, w=1")

axs[0].set_title("Varying Amplitude")
axs[0].set_xlabel("x")
axs[0].set_ylabel("y")
axs[0].legend()

for w in frequency:
    y = make_sine_wave(x, 1, w) 
    axs[1].plot(x, y, label=f"A=1, w={w}")

axs[1].set_title("Varying Frequency")
axs[1].set_xlabel("x")
axs[1].set_ylabel("y")
axs[1].legend()

for A, w in zip(amplitude, frequency): 
    y = make_sine_wave(x, A, w)
    axs[2].plot(x, y, label=f"A={A}, w={w}")

axs[2].set_title("Varying Amplitude and Frequency")
axs[2].set_xlabel("x")
axs[2].set_ylabel("y")
axs[2].legend()

plt.show()

#Bonus
axs[2].clear()

linestyles = ['-', '--', '-.', ':', (0, (5, 1))]
for (A, w), ls in zip(zip(amplitude, frequency), linestyles):
    y = make_sine_wave(x, A, w)
    axs[2].plot(x, y, linestyle=ls, label=f"A={A}, w={w}")

axs[2].set_title("Varying Amplitude and Frequency")
axs[2].set_xlabel("x")
axs[2].set_ylabel("y")
axs[2].legend()

plt.show()
```
