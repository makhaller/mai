```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)

data.shape
data[:5]
```

```python
print(data.shape)
magnitude = data[:, 0]
temperature = data[:, 1]

print("Min magnitude:", np.min(magnitude))
print("Max magnitude:", np.max(magnitude))

print("Min temperature:", np.min(temperature))
print("Max temperature:", np.max(temperature))

brightest = np.min(magnitude)
dimmest = np.max(magnitude)

print("Brightest star magnitude:", brightest)
print("Dimmest star magnitude:", dimmest)
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, label="Stars")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

brightest_index = np.argmin(magnitude)
dimmest_index = np.argmax(magnitude)
hottest_index = np.argmax(temperature)
coldest_index = np.argmin(temperature)

plt.scatter(log_temperature[brightest_index], 
            log_magnitude[brightest_index], 
            color="black", marker="*", s=300, label="Brightest")

plt.scatter(log_temperature[dimmest_index], 
            log_magnitude[dimmest_index], 
            color="purple", marker="*", s=300, label="Dimmest")

plt.scatter(log_temperature[hottest_index],
            log_magnitude[hottest_index], 
            color="green", marker="*", s=300, label="Hottest")

plt.scatter(log_temperature[coldest_index], 
            log_magnitude[coldest_index], 
            color="orange", marker="*", s=300, label="Coldest")

plt.legend()

plt.title("Hertzsprung-Russell Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

#Bonus
plt.scatter(log_temperature, log_magnitude, c=temperature, cmap='plasma')
plt.colorbar(label="Temperature")
```
